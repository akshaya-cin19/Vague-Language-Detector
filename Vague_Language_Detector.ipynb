```python

# Vague Language Detector

import re

vague_phrases = {
    "as soon as possible": "Specify an exact deadline.",
    "some things": "Mention the exact items or tasks.",
    "a lot": "Provide a specific number or quantity.",
    "soon": "Mention a date or time.",
    "somehow": "Explain the method clearly.",
    "good": "Describe what makes it good.",
    "bad": "Explain the specific problem.",
    "etc": "List the remaining important items.",
    "later": "Specify when the task should be done.",
    "maybe": "Clarify whether this is confirmed or uncertain."
}

def detect_vague_language(text):
    found = []

    for phrase, suggestion in vague_phrases.items():
        pattern = r"\b" + re.escape(phrase) + r"\b"

        if re.search(pattern, text, re.IGNORECASE):
            found.append((phrase, suggestion))

    print("=" * 45)
    print("VAGUE LANGUAGE DETECTOR")
    print("=" * 45)
    print("Input:", text)

    if found:
        print("\nVague phrases detected:")

        for phrase, suggestion in found:
            print(f"- {phrase}: {suggestion}")

        print(f"\nTotal vague phrases: {len(found)}")
    else:
        print("\nNo predefined vague phrases detected.")

    print("=" * 45)


# Sample input
text = (
    "Please finish this soon, complete some things, "
    "and send the report as soon as possible."
)

detect_vague_language(text)
```
